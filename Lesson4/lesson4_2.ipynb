```python
if True:
    print("這是單項選擇")
    print("這是True的程式區塊")
```

雙項選擇

```python
if False:
    print("這是雙項選擇")
    print("這是True的程式區塊")
else:
    print("這是雙項選擇")
    print("這是False的程式區塊")
```

學生總分為300
有些學生可以加分5%

```python
scores = int(input("請輸入學生分數(最高300分):"))
is_add = input("學生是否符合加分條件?(y,n)")
if (is_add == "y"):
    scores *= 1.05
    if scores > 300:
        scores = 300
    
print(f"學生分數是:{round(scores)}")
```

**多個except**

```python
divitionBy = 0 
try:
    print(30/"10")
    print(25 / divitionBy)
except ZeroDivisionError :
    print("除數不能為0")
except:
    print("發生不明錯誤")

    print("程式結束")
```

```python
try:
    scores = int(input("請輸入學生分數(最高300分):"))
    is_add = input("學生是否符合加分條件?(y,n)")
    if (is_add == "y"):
        scores *= 1.05
        if scores > 300:
            scores = 300

    print(f"學生分數是:{round(scores)}")
except:
    print("發生不知名錯誤!")
print("應用程式結束")
```
